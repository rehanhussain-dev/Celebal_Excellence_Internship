````python
def safe_divide(a, b):
    # YOUR CODE HERE
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Inputs must be numeric")
    try:
        return a / b
    except ZeroDivisionError:
        return None
    pass

# Tests
assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False, "Should have raised TypeError"
except TypeError as e:
    assert str(e) == "Inputs must be numeric"
print("1.3 passed ✓")
````

**Your Geometric Explanation (4.3):**

> Geometrically, an eigenvector is a special direction that does not change when a matrix transforms it. When a matrix acts on an eigenvector, the vector may get stretched, shrunk, or flipped, but it continues to point along the same line. Instead of undergoing a complicated transformation, the vector is simply multiplied by a number called the eigenvalue.

>Stretching (λ>1): When an eigenvector has an eigenvalue greater than 1, such as 5, the matrix makes the vector longer while keeping it in the same direction. Geometrically, this means the space is being expanded along that particular direction.

>Squishing (0< λ< 1): When an eigenvector has an eigenvalue between 0 and 1, such as 0.5, the matrix makes the vector shorter while keeping it in the same direction. Geometrically, this means the space is being compressed or contracted along that particular direction.

### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=False)  # YOUR CODE HERE

X_reconstructed = U @ np.diag(S) @ Vt  # YOUR CODE HERE — reconstruct from U, S, Vt
X_approx        = U[:, :1] @ np.diag(S[:1]) @ Vt[:1, :]  # YOUR CODE HERE — rank-1 approximation

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

**SVD → PCA Connection (4.4):**

> Mathematically, SVD factors an $m \times n$ matrix $X$ into:$$X = U S V^T$$

1. $U$ (Left Singular Vectors)

>What it is: An $m \times m$ orthogonal matrix.

>The Data Meaning: The columns of $U$ represent the structural relationships between your rows (the samples/observations).

>PCA Relation: They represent the coordinates of your original data points projected onto the new principal component axes, standardized to have unit variance.

2. $S$ (Singular Values)

>What it is: An $m \times n$ diagonal matrix containing non-negative values sorted from largest to smallest.

>The Data Meaning: These numbers represent the "strength" or magnitude of each structural axis.

>PCA Relation: They are directly related to the explained variance. The larger the singular value, the more information that axis captures. You can convert a singular value ($\sigma$) to a PCA eigenvalue ($\lambda$) using the formula:$$\lambda = \frac{\sigma^2}{n-1}$$

3. $V^T$ (Right Singular Vectors)

>What it is: An $n \times n$ orthogonal matrix (where $V^T$ is the transpose of $V$).

>The Data Meaning: The columns of $V$ (rows of $V^T$) represent the structural relationships between your columns (the features/variables).

>The matrix $V$ (or the rows of $V^T$) contains the directions of maximum variance.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = stats.kstest(s1, 'norm')  # YOUR CODE HERE
ks_s2 = stats.kstest(s2, 'norm')  # YOUR CODE HERE
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller(ts)  # YOUR CODE HERE — adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff = np.diff(ts)  # YOUR CODE HERE — np.diff
adf_diff = adfuller(ts_diff)  # YOUR CODE HERE
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    # YOUR CODE HERE
    combined = np.concatenate([expected, actual])
    bin_edges = np.linspace(np.min(combined) - 1e-5, np.max(combined) + 1e-5, bins + 1)
    expected_counts, _ = np.histogram(expected, bins=bin_edges)
    actual_counts, _ = np.histogram(actual, bins=bin_edges)
    expected_pct = expected_counts / len(expected)
    actual_pct = actual_counts / len(actual)
    epsilon = 1e-10
    expected_pct = np.where(expected_pct == 0, epsilon, expected_pct)
    actual_pct = np.where(actual_pct == 0, epsilon, actual_pct)
    psi = np.sum((actual_pct - expected_pct) * np.log(actual_pct / expected_pct))
    return psi
    pass

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)    # training distribution
drift_dist = np.random.normal(65, 12, 1000)    # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
# YOUR CODE HERE
sns.histplot(train_dist, color='navy', label='Train (Expected)', alpha=0.5, bins=15, ax=ax, kde=True)
sns.histplot(drift_dist, color='darkorange', label='Production (Actual)', alpha=0.5, bins=15, ax=ax, kde=True)
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.tight_layout(); plt.show()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

* **Concept Drift:** A change in how the input features relate to the target variable over time. This means the same input data may now produce different outcomes than it did during training. For example, changes in the economy may change what makes a borrower "risky."

* **Covariate Drift:** A change in the distribution of the model's **input features** over time, while the relationship between inputs and outputs remains the same. The model receives different kinds of data than it was trained on. For example, a marketing campaign may attract a younger group of users.

* **PSI < 0.1:** **Stable / No Drift.** The production data is very similar to the training data. No action is needed.

* **PSI 0.1–0.2:** **Moderate Drift.** The data distribution has changed slightly. The feature should be monitored and investigated.

* **PSI > 0.2:** **Significant Drift.** The production data is noticeably different from the training data. The model's performance may be affected, so immediate attention is required.

* **Retraining trigger example:** An automatic retraining process starts when an important feature (such as `income` or `credit_score`) reaches a **PSI of $\ge 0.25$ for a rolling 14-day period**, or when the model's production F1-score falls below a target threshold (for example, **0.80**) for **three consecutive days**.

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = (p_free_given_spam * p_spam) + (p_free_given_ham * p_ham)  # YOUR CODE HERE

# Bayes
p_spam_given_free = (p_free_given_spam * p_spam) / p_free  # YOUR CODE HERE

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    # YOUR CODE HERE
    prior_ham = 1.0 - prior_spam
    numerator = p_word_given_spam * prior_spam
    evidence = numerator + (p_word_given_ham * prior_ham)
    return numerator / evidence
    pass

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

**Bayes Term Mapping (6.3):**

| Term | Formula | In this example |
|------|---------|-----------------|
| Prior | P(Spam) | $0.30$ ($30\%$): The baseline probability that an incoming email is spam before looking at any words inside it. |
| Likelihood | P(word \| Spam) | $0.80$ ($80\%$): The probability that the word "free" appears, given that we already know the email is spam. |
| Evidence | P(word) | $0.275$ ($27.5\%$): The total probability of the word "free" appearing across all emails (spam and non-spam combined). |
| Posterior | P(Spam \| word) | $0.8727$ ($87.27\%$): The updated probability that the email is actually spam after we discover the word "free" inside it. |

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 5000
sample_size = 30
sample_means = np.array([np.random.choice(population, size=sample_size, replace=False).mean() for _ in range(n_samples)])  # YOUR CODE HERE — array of 5000 sample means

pop_mean = population.mean()
pop_std  = population.std()
clt_std  = pop_std / np.sqrt(sample_size)  # YOUR CODE HERE — predicted std of sample means

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()
# YOUR CODE HERE — histogram of sample means + theoretical normal overlay
ax.hist(sample_means, bins=45, density=True, alpha=0.6, color='skyblue', edgecolor='white', label='Sample Means')
x_axis = np.linspace(pop_mean - 4*clt_std, pop_mean + 4*clt_std, 200)
theoretical_pdf = stats.norm.pdf(x_axis, loc=pop_mean, scale=clt_std)
ax.plot(x_axis, theoretical_pdf, color='crimson', lw=2, label='CLT Theoretical Normal')
ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout(); plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

**CLT Reflection (6.4):**

> The Central Limit Theorem (CLT) is important because it helps us make reliable conclusions about a population even when we do not know the population's actual data distribution. It states that if we take many large random samples and calculate their averages, those averages will tend to follow a normal (bell-shaped) distribution. This idea forms the basis of many statistical methods, such as t-tests and ANOVA, and is also useful in machine learning techniques like Ordinary Least Squares (OLS) regression. As a result, even if the original data is highly skewed, the averages of large samples can still be treated as normally distributed.
